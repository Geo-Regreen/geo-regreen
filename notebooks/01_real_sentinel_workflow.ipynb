```python
import os
import ee
from dotenv import load_dotenv

# Load credentials and initialize Earth Engine
load_dotenv()
project_id = os.getenv("GEE_PROJECT")
ee.Initialize(project=project_id)

# 1. Define your Area of Interest - Johannesburg test
aoi = ee.Geometry.Point([28.0473, -26.2041]).buffer(10000) # 10km around JHB

# 2. Get Sentinel-2 imagery (renamed to 'image' since .first() gives a single image)
image = ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED') \
.filterBounds(aoi) \
.filterDate('2024-01-01', '2024-12-31') \
.filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 10)) \
.sort('CLOUDY_PIXEL_PERCENTAGE') \
.first()

# 3. Calculate NDVI - greenness
ndvi = image.normalizedDifference(['B8', 'B4']).rename('NDVI')

print("Got image:", image.get('system:index').getInfo())
print("NDVI ready. Generating thumbnail URL...")

# 4. Generate a viewable thumbnail URL (bypasses the widget renderer error)
thumb_url = ndvi.clip(aoi).getThumbURL({
'min': -0.2,
'max': 0.8,
'dimensions': 768,
'palette': ['white', 'yellow', 'green', 'darkgreen']
})

print("\nClick the link below to view your NDVI map:")
print(thumb_url)
```
